## Time-frequency analysis: power and inter-trial coherence

We now compute time-frequency representations (TFRs) from our Epochs.
We'll look at power and inter-trial coherence (ITC).

To this we'll use the function [`mne.time_frequency.tfr_morlet`](https://mne.tools/stable/generated/mne.time_frequency.tfr_morlet.html)
but you can also use [`mne.time_frequency.tfr_multitaper`](https://mne.tools/stable/generated/mne.time_frequency.tfr_multitaper.html)
or [`mne.time_frequency.tfr_stockwell`](https://mne.tools/stable/generated/mne.time_frequency.tfr_stockwell.html).

````python
# define frequencies of interest (log-spaced)
freqs = np.logspace(*np.log10([5, 30]), num=20)

# define number of cycles
n_cycles = 3

# decimate the signal by a factor of 3 for some additional speed
decim = 3

# compute the time-frequency representation
power, itc = mne.time_frequency.tfr_morlet(epochs, freqs=freqs, n_cycles=3, decim=decim, n_jobs=1)
````

````python
 # crop to remove edge artifacts
power.crop(-0.45, 1.6) 
itc.crop(-0.45, 1.6) 
````

### Inspect power

Let's look at the time-frequency spectra of all channels. When looking at time-frequency data, it is usually necessary to normalize the data by the baseline activity. There are many ways to do this, we will use the "meanlogratio" method by [Kinley et al. (2026)](https://doi.org/10.1016/j.jneumeth.2026.110826) in this lesson.

````python
# Some setting for our baseline, which will be applied to the plots
baseline_mode = "meanlogratio"
baseline = (-0.45, -0.1)
````

```{warning}

The baseline we work with is way too short for the low frequencies in our decomposition:
For 5 Hz, or time window of 3 cycles are 600 ms long. This is already longer than our baseline!
Plus, we chose only 100 ms of buffer towards 0: If we center the 5 Hz - 3 cycles wavelet at -100 ms, we reach 200 ms beyond 0, thus we will leak stimulus activity into our baseline: not good.

So, what to do?
In real life, we would re-cut the epochs to get more than 500 ms of baseline!
```

````python
power.plot_topo(baseline=baseline, mode=baseline_mode, title="Average power");
````

```{note}
The generated figure is interactive. In the topo you can click on an image to visualize the data for one sensor. You can also select a portion in the time-frequency plane to obtain a topomap for a certain time-frequency region.

```

````python
power.plot([83], baseline=baseline, mode=baseline_mode, title=power.ch_names[83]);
````

### Compute FWHM for our wavelets

We defined our wavelets based on _number of cycles_. Mike X Cohen (Cohen 2019, NeuroImage (199, p. 81-86)) argues that it is better to define the full-width-at-half-maximum (FWHM) of the Morlet wavelet.

You might remember that our wavelet of `n_cycles` gets multiplied with a Gaussian taper that dampens the wavelet towards the edges. Thus, the FWHM is a better estimate of the temporal smoothing than the full length of `n_cycles`.

Let's see how to do this:

````python
# First let's check our cycles per frequency - we had specified them above:
freqs, n_cycles  # this is the number of cycles we asked for per frequency
````

The formula to retrieve the FWHM is (Cohen 2019, eq. 4):

$ FWHM = \frac{n  \sqrt(2  \ln 2)}{\pi * f} \enspace,$

where $n$ is the number of cycles, $f$ is the frequency, and $\ln$ denotes the natural logarithm.

Let's convert this equation into a little function!

````python
def get_fwhm_morlet(n_cycles, freq):
    """Estimate the FWHM of a Morlet wavelet."""

    fwhm = (n_cycles * np.sqrt(2 * np.log(2))) / (np.pi * freq)
    return fwhm
````

Now let's estimate what length of the FWHM of our Morlet wavelets was in seconds:

````python
for freq in freqs:
    
    # estimate fwhm
    fwhm = get_fwhm_morlet(n_cycles, freq)

    # print it
    print(f"FWHM at {freq:.3f} Hz was {fwhm:.3f} s")
````

## Evoked power

Up to now, we have been looking at total power: compute PSD or TFR for each epoch and then average the PSDs/TFRs across epochs. We can also look at evoked power and compute the PSD or TFR for the evoked response:

````python
evoked = epochs["face"].average()
evoked_power = evoked.compute_psd()
evoked_power.plot_topomap();
````
